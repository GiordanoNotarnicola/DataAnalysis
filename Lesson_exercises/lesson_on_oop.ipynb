```python
import numpy as np
class Vector2d:

    def __init__(self, x: float, y:float):
        self.x=float(x)
        self.y=float(y)

    def norm(self):
        return np.sqrt(self.x**2+self.y**2)

    def r(self):
        return self.norm

    def phi(self):
        return np.atan2(self.y, self.x)

    def dot(self, other: "Vector2d"):
        return self.x*other.x + self.y*other.y

    def __str__(self):
        return f"{self.__class__.__name__}({self.x}, {self.y})"
        
    

if __name__=="__main__": #a way to don't execute the script if you import that. 
    n1=Vector2d(1.,0.)
    print(n1)
    n2= Vector2d(0.,1.)
    print(n2.norm())
    print(n1.dot(n2))
```

Now I explain encapsulation: You have to separate the public interface and the private states, that it will change to upgrade the script. So you have to give the methods and not the state that you __init__, so the user can have only the method that do forever the same thing. you hide the variable with a leading underscore and the define them. gentlemen agreement that say you can't use underscore.

```python
import numpy as np
class Vector2d:

    def __init__(self, x: float, y:float):
        self._x=float(x)
        self._y=float(y)

    def x(self):
        return self._x

    def y(self):
        return self._y


    def norm(self):
        return np.sqrt(self._x**2+self._y**2)

    def r(self):
        return self.norm

    def phi(self):
        return np.atan2(self._y, self._x)
```

Inheritance is a relation between the base class and the daughter class: the daughter has the same method and state fo the base class. In this case, all the beahviour is in the base classe and the children only change the states. Overloading is overwriting the same behaviour in child classess

```python
class Particle:

    def __init__(self, mass, charge=0, name=None, momentum=0):
        self.mass= mass
        self.charge= charge
        self.name= name
        self.momentum=momentum

    def energy(self):
        return np.sqrt(self.momentum**2+ self.mass**2)



class Electron(Particle):

    def __init__(self, momentum=0):
        Particle.__init__(self, 0.51, -1., "e", momentum)


el=Electron(momentum=3)
print(f"Energy of {el.name}, with {el.charge} charge is {el.energy():.4} Mev/c^2")
       
```
