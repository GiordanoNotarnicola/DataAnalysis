we want to change x, and the user want to acces x and y, even that they don't exist (we set r and phi)

```python
import math

class Vector2d:

    def __str__(self):
        return f"Vector2d(r={self.r}, phi={self.phi})"
    
    def __init__(self, r, phi):
        self._r = float(r)
        self._phi = float(phi)
    @property #fingo che r non sia un metodo, ma un attributo, dunque posso chiamarlo senza parentesi
    #faccio encapsulation senza creare getter and setters
    def r(self):
        return self._r

    @property
    def phi(self):
        return self._phi

    @r.setter #property per fare sì che diventi un setter
    def r(self, new_r):
        self._r = float(new_r)

    @phi.setter #property per fare sì che diventi un setter
    def phi(self, new_phi):
        self._phi = float(new_phi)


    @property
    def x(self):
        return self._r*math.cos(self._phi)

    @property
    def y(self):
        return self._r*math.sin(self._phi)

    @x.setter
    def x(self, x):
        new_r=math.sqrt(self.y**2+x**2)
        new_phi= math.atan2(self.y, x)
        self._phi=new_phi
        self._r=new_r

    @y.setter
    def y(self, y):
        new_r=math.sqrt(y**2+self.x**2)
        new_phi= math.atan2(y, self.x)
        self._phi=new_phi
        self._r=new_r

v=Vector2d(3., 0.5*math.pi)
print(v.r)
v.r=5.
v.y=1.
v.phi=0.1
print(v)
type(v.r)
```

lets' find special methods, __repr__, __str__. the compiler prefer str, if there's not use repr, so we usually use str. __add__, __mul__, __rmul__ and other special method. we have two different type of multiplication per a scalar because we have to ordinate the factors.

```python
import math

class Vector2d:
    def __init__(self, x, y):
        self.x = float(x)
        self.y = float(y)

    def __str__(self):#automatically return a string of the istance
        return f"(x={self.x}, y={self.y})"

    def __repr__(self):
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        return Vector2d(self.x +other.x, self.y+ other.y)

    def __mul__(self, scalar):#by a scalar
        return Vector2d(scalar*self.x, scalar*self.y)

    def __rmul__(self, scalar):
        return self*scalar

    

v= Vector2d(3., -1.)
z= Vector2d(2., 5.)
print(v+z)
print(3*v)
print(z*3)

    
```

N-array, we want an array syntax( v[0], ecc), we find more magical methods

```python
import math
from array import array

class Vector:

    TYPE_CODE= "d"

    def __init__(self, components):
        self._components = array(self.TYPE_CODE, components)

    def __str__(self):
        return (str(self._components))

    def __repr__(self): #a way to adapt graphically what I want, output is a string
        components = str(self._components)
        components = components[components.find("["): -1]
        return f"{self.__class__.__name__}({components})"

    def __getitem__(self, index): #get to the istance a structure array like, so v[i], call a float
        return self._components[index] 

    def __setitem__(self, index, value):#useful to change an element in a natural way in an array: v[index]=value
        self._components[index]=value

    

v=Vector([1., 1., 2., 4.])
print(v)
v[0]=3.
print(v[0]+v[2])
type(v[0])

```

in python you don't have to specify the type, if the class has the correct method, you can use it: the istance of a class is not iterable, but with the method __iter__ i can trait it as an iterable object with all those functions.

```python
import math
from array import array

class Vector:

    TYPE_CODE= "d"

    def __init__(self, components):
        self._components = array(self.TYPE_CODE, components)

    def __iter__(self):
        return iter(self._components)

v=Vector([1., 3., 5., 2.])
print(type(v))
for i in v:
    print(i)
```

last but non least... __call__ method let an object callable. 

```python
class Line:
    def __init__(self, slope=1., intercept=0.):
        self.slope= slope
        self.intercept= intercept

    def __call__(self, x):
        return self.slope*x + self.intercept
```
