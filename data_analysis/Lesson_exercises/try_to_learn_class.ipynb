Esempio di classe con sole funzioni

```python
class Television:
    def turn_on(self, channel):
        print(f"Turning on {self}")
        print(f"Showing channel {channel}")

    def turn_off(self):
        print(f"Turning off {self}")

mytv=Television()
yourtv=Television()
mytv.turn_on(4)
yourtv.turn_off()
```

Esempio di classe con __init__

```python
class Television:

    def __init__(self, owner):
        print("Creating a television istance")
        self.model="Lg 4k-oled"
        self.owner= owner

    def info(self, poll):
        print(f"this television model is {self.model}, owned by {self.owner}, is {poll} thumbs.")

mytv=Television("giordano")
mytv.info(27)
```
