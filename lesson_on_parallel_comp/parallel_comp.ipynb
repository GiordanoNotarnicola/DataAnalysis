Threads: memory assigned to the programm, then you have some threads that have their local memory, with code and stacks. you can open two process and they run in two different memory spaces, but the threads run in the same memory space.
be careful about:
- starvation: if you have several task and are uno dopo l'altra, se non puoi aprire il file iniziale, il task non può terminare.
- deadlock: il primo task dipende da risultato del secondo e viceversa... dunque non ne puoi uscire.
to have a thread-safe, the could share a counter, but it can count wrong because the process are in parallel: if i do read 1 and i want to add 2 and do the same in the other thread, sequentially, then i obtain 3 in both and not 3 and 5. we don't want that.
use mutex threading.Lock(), to wait the other thread to operate and then ready to go.
use semaphore are a generalization of mutex. 
in python you have GIL (concurrentially), global interpreter lock: if you have two different threads, python cannot run the two threads at the same time, so is single-threading. 
if: run->i/o->run->i/o you can open file and start computation.
