Algo to find maximun value in a list

```python
a=[1, 3, 4, 3, 6, 7, 9, 12, 6]
def f_max(values:list)->int:
    count=0
    maxi=values[0]
    for i in values:
        if maxi<i:
            maxi=i
        count+=1
    return maxi, count

f_max(a)
```

Algo to find a specific number in a list

```python
a=[1, 3, 4, 3, 6, 7, 8, 9, 12, 6]
def seq_search(lista:list, n:int)->str:
    passi=0
    for i, val in enumerate(lista):
        passi+=1
        if val==n:
            return f"{n} è presente in posizione {i}, numero passi {passi}"
    return f"{n} non è presente in lista"
seq_search(a, 9)
```

Binary research algo, se la lista è ordinata allora posso aprire a metà la lista: se il mio valore è minore di quello cercato, prendo la parte destra e viceversa.

```python

def binary_search(values: list, target: int) -> int:
    left = 0
    right = len(values) - 1
    count=0
    while left <= right:
        count+=1
        midpoint = (left + right) // 2
        value = values[midpoint]
        if value == target:
            return midpoint, count
        elif value < target:
            left = midpoint + 1
        else:
            right = midpoint - 1
    return -1, count

l = [1, 2, 4, 6, 12, 32, 99, 100, 150, 173]
print(binary_search(l, 4))
print(binary_search(l, 3))
```

Some pre-created algos

```python
import bisect

def binary_search(values: list, target: int) -> int:
    i = bisect.bisect_left(values, target)
    if values[i] == target:
        return i
    return -1

l = [1, 2, 4, 6, 12, 32, 99]
print(binary_search(l, 4))
print(binary_search(l, 3))
```

```python
import bisect

def slightly_better_sort(values: list) -> list:
    sorted_list = []
    for value in values:
        i = bisect.bisect_left(sorted_list, value)
        sorted_list.insert(i, value)
    return sorted_list

l = [10, 1, 8, 5, 2, 7, 3, 9, 4, 6]
print(slightly_better_sort(l))
```
