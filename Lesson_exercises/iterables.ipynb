differences between iterators and generators is that the first has a container already existing with data, the second create the container on the fly and put in data. this is called lazy generations. how to create an iterable object?
- we create iterable object such as dictionaries, tuples, lists, arrays. an iterable is a container + an__iter__() method. all the python built-in container has an iter method, but if we create our own class, we need it. the__iter__() method return iterator: it points an object and has a__next__() method: self.index=0, then point, then__next__() and so on. when container has no more elements start an exception (inheritated) StopIteration().
the beauty of__next__() method is that you can iterate in which way you want.

myiterator= mylist.iter()
while True:
    try:
        number=myiterator.next()
        print(number)
    except StopIteration():
        break

- we create a generator through a generator expressions. in python3, range is a generator. 

```python
for i in range(4):#generator
    print(i)


data=[1, 3, 4]
square_data_gener=(x**2 for x in data)#generator expressions
for square_datum in square_data_gener:
    print(square_datum)
```

```python
def g_f():
    print('first')
    yield 1
    print('second')
    yield 2
    print('StopIteration exception')

gen=g_f()
print(next(gen))
print(next(gen))
next(gen)
```

```python
#generator of infinite fibonacci number

def fibo():
    a,b=0,1
    while True:
        yield a
        a,b=b, a+b

max_n=7
fib_num=[]
for i, fib in enumerate(fibo()):
    if i>= max_n:
        break
    else:
        fib_num.append(fib)
print(fib_num)
#or we can call a module

import itertools#other way to study
fib_gen=(fib for fib in itertools.islice(fibo(), max_n))
print(list(fib_gen))
```

lambda functions is one line functions with no name

```python
mult=lambda x, y :x*y #lambda, list of argument comma-separate, semicolon and the operation
print(mult(2, -3))

numbers=range(10)
squares=list(map(lambda n: n**2, numbers))
print(squares)

```
