functions in python are first class object, so you can pass it as argument of another function.

```python
def outer():
    def inner():
        return
    return inner

my_func=outer()
print(my_func.__name__)
my_func()

def outer2():
    string='hello'
    def inner():
        print(string)
    return inner

other_func=outer2()
other_func()
```

use a function in one other is really efficient to store less data. it is very common, so we use decorators

```python
def mult(a,b):
    print('execute {} x {}'.format(a, b))
    return a*b

def add_n_wrapper(func,n):

    def wrapper(*args, **kwargs):
        result=func(*args, **kwargs)#coming from the external layer
        print('add {}'.format(n))
        return result+n

    return wrapper

func_plus_five=add_n_wrapper(mult,5)
print('result= {}'.format(func_plus_five(2,3)))

#it's very difficult...
```

```python
def print_funct_info(func):
    def wrapper(*args, **kwargs):
        print('Calling function \'{}\''.format(func.__name__))

        return func(*args, **kwargs)
    return wrapper

@print_funct_info #the decorator
def some_f(a, b, c=0):
    return a*b+c
```
